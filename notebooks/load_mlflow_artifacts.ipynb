```python
def test_load_production_model_and_preprocessor(model_name, alias):

    client = MlflowClient()

  

    version_details = client.get_model_version_by_alias(
        model_name,
        alias,
    )
    print(version_details )
    print(version_details.source)
    
    model_uri = f"models:/{model_name}@{alias}"
    print("loading...model")
    
    model = mlflow.sklearn.load_model(model_uri)
    print("loading preprocessor")
    preprocessor_path = mlflow.artifacts.download_artifacts(
        artifact_uri=f"{version_details.source}/artifacts/preprocessor.joblib"
    )

    preprocessor = joblib.load(preprocessor_path)
    
    
    return model , preprocessor , preprocessor , version_details
```
